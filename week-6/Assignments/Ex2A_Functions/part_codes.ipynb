```python
def get_supplier_code(name):

  for characters in range(len(name)):
    if name[characters] == ':' :
      print(name [:characters])
```

```python
def get_product_num(name):

  for characters in range(len(name)):
    if name[characters] == ':':
      begin = name[characters+1:]
      for characterz in range(len(begin)):
        if begin[characterz] == '-':
          end = begin[:characterz]
          print(end)
```

```python
def get_size(name):

  for characters in range(len(name)):
    if name[characters] == '-' :
      print(name [characters+1:])
```

```python
Acme = "acme:123-L"
DI = "di:12345-M"
Ace = "ace:1-12"

get_supplier_code(Acme)
get_supplier_code(DI)
get_supplier_code(Ace)

get_product_num(Acme)
get_product_num(DI)
get_product_num(Ace)

get_size(Acme)
get_size(DI)
get_size(Ace)

```

```python
def display_part_info(name):
    supplier = name.split(":")[0]
    product  = name.split(":")[1].split("-")[0]
    size     = name.split("-")[1]
    print(f"For {name}, the supplier is {supplier}. Part number {product} is available in size {size}.")

display_part_info(Acme)
display_part_info(DI)
display_part_info(Ace)
```
