Lab 2 of Exercise 2.B, Working with datetime functions

```python
import datetime
```

#Today

```python
today = datetime.datetime.now()
print(today)
```

```python
print(today.strftime("%A, %B, %D, %Y"))
```

# My birthday

```python
birthday = datetime.date(1998, 7, 24)
```

```python
print(birthday.strftime("%m/%d/%Y"))
```

```python
print(f"{birthday.strftime("%m/%d/%Y")} is my birthday")
```

#   Next 90 Days 

```python
ninety_d = today + datetime.timedelta(days=90)
```

```python
print(ninety_d.strftime("%m/%d/%Y"))
```

```python
print(f"90 days from now is {ninety_d.strftime("%m/%d/%Y")}")
```

# Time for Dinner 

```python
dinner_time = datetime.time(7, 20)
```

```python
print(dinner_time.strftime("%H:%M"))
```

```python
print(f"Let's meet for dinner at {dinner_time.strftime("%H:%M")} PM")
```
