# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop iterates over the rows of arrA (each 'x' is a 1D array/row), so multiplying by 2 doubles each entire row at once. The second loop iterates over individual elements within each row, so it prints each single number multiplied by 2.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2  # x*2 multiplies the whole row, not just the individual element y
print(loop_result)
# loop_result ends up as the last row of arrA multiplied by 2, because x is reassigned
# on each outer iteration. The inner loop runs but only x*2 (entire row) is stored each time.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
# stacked_array_2 = np.stack((joined_array_1, joined_array_2))
# This raises a ValueError: all input arrays must have the same shape.
# joined_array_1 is (4,3) and joined_array_2 is (2,6) — shapes don't match, so stack fails.
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
# another_2 = np.concatenate((joined_array_1, arrA), axis=1)
# This raises a ValueError: all the input array dimensions except for the concatenation axis must match exactly.
# joined_array_1 has shape (4,3) and arrA has shape (2,3); they have different numbers of rows,
# so you can't concatenate them along axis=1.
```

```python
hstacked = np.hstack((arrA, arrA))   # stacks side by side (along columns)
print('hstack:', hstacked)

vstacked = np.vstack((arrA, arrA))   # stacks top to bottom (along rows)
print('vstack:', vstacked)

dstacked = np.dstack((arrA, arrA))   # stacks along the third (depth) axis
print('dstack:', dstacked)
```

## Splitting Arrays

```python
splitsky = np.split(joined_array_1, 2)
print(splitsky)
# Result: 2 arrays, each with 2 rows and 3 columns (2D arrays)
```

```python
spliterino = np.split(joined_array_1, 3) if joined_array_1.shape[0] % 3 == 0 else 'Cannot split evenly into 3'
print(spliterino)
# joined_array_1 has 4 rows; 4 is not divisible by 3, so this raises a ValueError.
# To demonstrate it working: np.split needs equal-sized splits unless you pass indices.
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7, 3]])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
