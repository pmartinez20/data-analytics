# Exercise 1.C Performing operations on arrays

```python
import numpy as np
```

## Sample Arrays

```python
r_1 = np.random.randint(1, 21, size=2)
r_2 = np.random.randint(1, 21, size=(1, 2))
r_3 = np.random.random(size=(2, 1))
r_4 = np.random.randint(20, 101, size=(4, 5))
r_5 = np.random.random(size=(4, 5))
```

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

shuffle() shuffled the rows of r_4 (it shuffles along the first axis), so the individual values within each row stayed together but the row order changed.

```python
new_shuffle = np.random.permutation(r_4)
print('new_shuffle:', new_shuffle)
print('r_4 unchanged:', r_4)
```

Use **shuffle** when you want to rearrange an array in-place (modifies the original). Use **permutation** when you want a new shuffled copy and need to keep the original array intact.

## Mathematical Operations

```python
print('r_1:', r_1, 'shape:', r_1.shape)
print('r_2:', r_2, 'shape:', r_2.shape)
result = np.add(r_1, r_2)
print('r_1 + r_2:', result, 'shape:', result.shape)
```

```python
print('r_1:', r_1, 'shape:', r_1.shape)
print('r_3:', r_3, 'shape:', r_3.shape)
result2 = np.add(r_1, r_3)
print('r_1 + r_3:', result2, 'shape:', result2.shape)
```

```python
print('r_2:', r_2, 'shape:', r_2.shape)
print('r_4:', r_4, 'shape:', r_4.shape)
result3 = np.add(r_2, r_4)
print('r_2 + r_4:', result3, 'shape:', result3.shape)
```

```python
# np.add(r_1, r_4) raises a ValueError: operands could not be broadcast together with shapes (2,) (4,5)
# r_1 has 2 elements, r_4 has 5 columns — incompatible for broadcasting.
# np.add(r_2, r_4) works because r_2 is (1,2) — but r_4 is (4,5), 2 != 5 so it also fails.
# np.add(r_3, r_4) fails too: r_3 is (2,1), r_4 is (4,5) — 2 rows vs 4 rows, can't broadcast.
print('Attempting r_1 + r_4 would raise: shapes (2,) and (4,5) not broadcastable')
```

```python
print(np.add(np.array([1]), r_4))  # broadcasts scalar-like [1] to every element
```

```python
print(np.add(np.array([[1]]), r_4))  # also broadcasts [[1]] to every element
```

```python
print(np.add(np.array([1,1,1,1,1]), r_4))  # adds 1 to each column across all rows
```

```python
# np.add(np.array([[1],[1]]), r_4)
# Raises ValueError: shapes (2,1) and (4,5) not broadcastable — 2 rows cannot broadcast to 4 rows.
```

**Broadcasting** lets NumPy add arrays of different shapes by stretching the smaller array along size-1 dimensions to match the larger one. When using add(), NumPy aligns dimensions from the right: if sizes match or one of them is 1, it broadcasts; otherwise it raises an error. The shape of the result matches the larger array's shape.

## Rounding

```python
print(r_3)
```

```python
print(np.trunc(r_3))   # truncates toward zero (drops decimal part)
```

```python
print(np.fix(r_3))     # same as trunc — rounds toward zero
```

```python
print(np.floor(r_3))   # rounds down to nearest integer
```

```python
print(np.ceil(r_3))    # rounds up to nearest integer
```

```python
print(np.round(r_3, 2))  # rounds to 2 decimal places
```

## Grocery Problem

```python
# Categories: [vegetables, fruits, snacks, beverages]
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])
print('Bag 1:', bag1)
print('Bag 2:', bag2)
```

```python
totals = np.array([np.sum(bag1), np.sum(bag2)])
print(f'The first bag contains {totals[0]} items, and the second bag contains {totals[1]} items.')
```

```python
combined = np.add(bag1, bag2)
print(f'There are {combined[0]} vegetables, {combined[1]} fruits, {combined[2]} boxes of snacks, and {combined[3]} beverages in the two grocery bags.')
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])
```

```python
# Ben missed Monday — assigned 0 for that day
ben = np.array([0, 5, 5, 3, 4])
```

```python
# Carla was sick Wednesday — assigned 0 for that day
carla = np.array([3, 6, 0, 3, 4])
```

```python
all_knocks = np.array([ali, ben, carla])
print(all_knocks)
```

```python
per_person = np.sum(all_knocks, axis=1)
group_total = np.sum(all_knocks)
print(f'Ali: {per_person[0]}, Ben: {per_person[1]}, Carla: {per_person[2]}')
print(f'Group total: {group_total}')
```

```python
daily_totals = np.sum(all_knocks, axis=0)
rolling_daily = np.cumsum(daily_totals)
estimated_votes = rolling_daily * 0.5
print('Rolling estimated votes by day:', estimated_votes)
```
