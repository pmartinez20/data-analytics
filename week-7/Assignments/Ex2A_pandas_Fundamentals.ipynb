```python
import numpy as np
import pandas as pd
```

## Series

```python
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {'first': 13, 'second': 14, 'third': 15, 'fourth': 16}
```

```python
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(ac)
ser4 = pd.Series(ad)
print(ser1)
print(ser2)
print(ser3)
print(ser4)
```

```python
# ser5 = pd.Series(17, 18, 19, 20)
# TypeError: __init__() takes from 1 to 2 positional arguments but 5 were given
# pd.Series() expects a single iterable, not multiple positional arguments.
```

## DataFrames from Series

```python
df1 = pd.DataFrame(ser1)
df1
```

```python
df1.insert(1, 'col2', ser2)
df1.insert(2, 'col3', ser3)
df1
```

```python
df1['col4'] = ser4
df1
# ser4 uses string index labels ('first','second','third','fourth') instead of integer positions.
# pandas tries to align on index, but df1 uses 0-3 integer index, so none match — all NaN values appear.
```

```python
df1.loc[4] = [5, 9, 12, 0]
df1
```

## Reading Data into a DataFrame

```python
df2 = pd.read_csv('sakila_example.csv')
df2
```

```python
df2.head(5)
```

```python
df2.tail(10)
```

```python
df2.info()
```

From the info results: the dataset has 5 columns (title, description, category, language, actor_name), all with object (string) dtype, and no null values. This tells us it's a movie/actor dataset with fully populated text fields.

```python
df2.describe()
```

The describe results show counts and unique value counts for each text column, plus the most frequent (top) value and its frequency. Useful: seeing which category or actor appears most often. Less useful: the raw counts and 'freq' for columns like 'description' where nearly every value is unique.

```python
df3 = pd.read_csv('northwind_example.csv')
df3
```

```python
df3.describe()
```

The northwind describe results include numeric statistics (mean, std, min, max, quartiles) for UnitPrice and Quantity because those columns are numeric, unlike sakila which was all text. This gives much more actionable insights — e.g., the average unit price, the range of quantities ordered.
