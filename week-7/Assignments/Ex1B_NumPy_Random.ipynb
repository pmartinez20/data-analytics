```python
import numpy as np
```

## Generating Random Numbers

```python
r_1 = np.random.randint(1, 21, size=2)
print(r_1)
```

```python
r_2 = np.random.randint(1, 21, size=(1, 2))
print(r_2)
```

```python
r_3 = np.random.random(size=(2, 1))
print(r_3)
```

```python
r_4 = np.random.randint(20, 101, size=(4, 5))
print(r_4)
```

```python
r_5 = np.random.random(size=(4, 5))
print(r_5)
```

## Generating Random Choices from an Array

```python
students = np.array(['Alice', 'Bob', 'Carmen', 'David', 'Eva', 'Frank',
                      'Grace', 'Henry', 'Iris', 'James', 'Karen', 'Leo'])
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)
# This would NOT be an effective way to assign breakout groups because it samples with
# replacement by default, so the same student could appear in multiple groups.
```

```python
# Probabilities assigned to each of the 12 students:
# Alice: 0.30 (highest — 30% chance), Bob: 0.00 (zero chance),
# Carmen: 0.10, David: 0.10, Eva: 0.10, Frank: 0.08,
# Grace: 0.08, Henry: 0.07, Iris: 0.05, James: 0.05,
# Karen: 0.04, Leo: 0.03 — all probabilities must sum to 1.0
p_values = [0.30, 0.00, 0.10, 0.10, 0.10, 0.08, 0.08, 0.07, 0.05, 0.05, 0.04, 0.03]
call_on_weighted = np.random.choice(students, size=5, p=p_values)
print(call_on_weighted)
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
